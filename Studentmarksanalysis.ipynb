```python
import pandas as pd

df = pd.read_csv("/archive (2).zip")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Total"]= df["number_courses"]+ df["time_study"]+ df["Marks"]
df['Average']= df["Total"]/3
df
```

```python
top_student = df.loc[df["Total"].idxmax()]
top_student
```

```python
subject_avg = df[["number_courses","time_study","Marks"]].mean()
subject_avg
```

```python
df["Marks"]= df["Average"].apply(lambda x: "Pass" if x >= 50 else "Pass","Fail")
df
```

```python
df["Marks"].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.bar(df["number_courses"],df["Average"])
plt.xlabel("Number of Courses")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```
